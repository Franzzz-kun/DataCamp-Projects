```python
import pandas as pd
import matplotlib.pyplot as plt
# import numpy as np

# Read in the Netflix CSV as a DataFrame
netflix_df = pd.read_csv("netflix_data.csv", index_col = 0)
df_new = netflix_df.loc[:, ["type", "duration", "release_year"]]

# For the filtering
pd.set_option('display.max_rows', None)
df_filter = df_new[(df_new['type'] == "Movie") & (df_new['release_year'] >= 1990) & (df_new['release_year'] <= 1999)]


duration_list = []
release_year_list = []


for i in df_filter['duration']:
    duration_list.append(i)
for j in df_filter['release_year']:
    release_year_list.append(j)

plt.hist(duration_list, bins = 10)
plt.title("Histogram for Most Frequent Movie Duration in the 1990s")
plt.xlabel("Duration")
plt.ylabel("Number of Films")
plt.show()

```
